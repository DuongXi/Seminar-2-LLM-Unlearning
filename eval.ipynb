Function to select 40 samples for Hallu Rate eval, remember to change MODEL & MASTER before running cells from this point onward

```python
import json, random

NEW_MASTER = "master.json"
src, dst, n = MASTER, NEW_MASTER, 40
random.seed(42)

with open(src) as f:
    rows = [json.loads(line) for line in f if line.strip()]

sample = random.sample(rows, n)

with open(dst, "w") as f:
    for r in sample:
        f.write(json.dumps(r) + "\n")
```
